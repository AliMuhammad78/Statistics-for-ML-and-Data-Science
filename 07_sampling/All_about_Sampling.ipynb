# Sampling in Statistics

> **Sampling** is the statistical process of selecting a smaller, representative group from a larger population to collect data and draw conclusions about the entire population.

---

## 1. Introduction

In statistics and data science, we are often interested in studying a **population** — the complete group of individuals, objects, events, or observations that we want to understand.

However, studying every member of a population is often impractical, expensive, time-consuming, or even impossible. Instead, we select a smaller group called a **sample** and use the information obtained from that sample to make conclusions about the population.

This process is known as **sampling**.

### Simple Example

Suppose a university has **20,000 students** and we want to know their average daily study time.

It would be difficult to ask every student.

Instead, we might randomly select **500 students**, collect their study-time data, calculate the average, and use that result to estimate the average study time of all 20,000 students.

```text
Population
20,000 Students
       |
       | Sampling
       v
Sample
500 Students
       |
       | Statistical Analysis
       v
Estimate / Conclusion
about the Population
```

---

# 2. Key Terms in Sampling

Understanding the terminology is essential before studying sampling methods.

| Term               | Meaning                                                                                   |
| ------------------ | ----------------------------------------------------------------------------------------- |
| **Population**     | The complete group being studied                                                          |
| **Sample**         | A smaller group selected from the population                                              |
| **Sampling**       | The process of selecting a sample                                                         |
| **Parameter**      | A numerical characteristic of a population                                                |
| **Statistic**      | A numerical characteristic calculated from a sample                                       |
| **Sampling Frame** | A list or source containing members of the population from which a sample can be selected |
| **Sampling Unit**  | The individual element selected during sampling                                           |
| **Census**         | Collecting data from every member of the population                                       |

### Example

Suppose we want to study the average GPA of students at a university.

* **Population:** All university students
* **Sample:** 500 selected students
* **Parameter:** True average GPA of all students
* **Statistic:** Average GPA calculated from the 500 students
* **Sampling Unit:** Individual student
* **Sampling Frame:** University student enrollment list

# 3. Population vs Sample

The distinction between population and sample is fundamental.

### Population

A **population** contains every member relevant to the research question.

Examples:

* All voters in a country
* All customers of a company
* All students in a university
* All houses in a city
* All transactions in a banking system

### Sample

A **sample** is a subset of the population selected for analysis.

```text
                 POPULATION
        ┌───────────────────────────┐
        │ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ │
        │ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ │
        │ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ │
        │ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ ○ │
        └───────────────────────────┘
                    ↓
                 SAMPLE
              ┌───────────┐
              │ ○ ○ ○ ○ ○ │
              │ ○ ○ ○ ○ ○ │
              └───────────┘
```

The objective is to select a sample that provides useful information about the population.

# 4. Why Do We Need Sampling?

Sampling is necessary because studying an entire population is frequently impractical.

## 4.1 Large Population Size

Some populations are extremely large.

For example:

* All internet users
* All customers of a multinational company
* All households in a country
* All transactions occurring on a large platform

Studying every individual may require enormous resources.

---

## 4.2 Lower Cost

Collecting information from an entire population can be expensive.

Sampling reduces:

* Survey costs
* Labor costs
* Transportation costs
* Data collection costs
* Processing costs

A well-designed sample can provide useful information at a fraction of the cost of a census.

---

## 4.3 Saves Time

Sampling allows researchers to obtain results much faster.

A census may take months or years, whereas a properly designed sample can sometimes be collected within days or weeks.

This is particularly important when decisions must be made quickly.

---

## 4.4 Sometimes a Census Is Impossible

In some situations, testing every item destroys the item being tested.

For example:

> If a manufacturer wants to test whether a light bulb lasts for 10,000 hours, testing every bulb until failure would destroy the products.

Sampling is therefore necessary.

---

## 4.5 Easier Data Management

A smaller dataset is generally easier to:

* Collect
* Clean
* Store
* Process
* Analyze
* Visualize
* Interpret

This becomes especially important in data science and machine learning.

# 7. What Makes a Good Sample?

A good sample should provide a reasonable representation of the population.

Important characteristics include:

### Representativeness

The sample should reflect important characteristics of the population.

### Randomness

Where appropriate, selection should involve randomness to reduce systematic selection bias.

### Adequate Sample Size

The sample should contain enough observations to produce reliable estimates.

### Low Bias

The sampling process should avoid systematically favoring particular members of the population.

### Appropriate Sampling Method

The method should match the research question, population structure, resources, and available sampling frame.

# 9. Major Types of Sampling

Sampling methods are broadly divided into two major categories:

```text
                         SAMPLING
                            |
             ┌──────────────┴──────────────┐
             |                             |
      PROBABILITY SAMPLING        NON-PROBABILITY SAMPLING
             |                             |
       Random Selection              Non-random Selection
```

## A. Probability Sampling

In probability sampling, members of the population have a **known, non-zero probability of being selected** according to a defined sampling design.

Major methods include:

1. Simple Random Sampling
2. Systematic Sampling
3. Stratified Sampling
4. Cluster Sampling
5. Multistage Sampling

---

## B. Non-Probability Sampling

In non-probability sampling, selection is not based on a known probability mechanism for every population member.

Major methods include:

1. Convenience Sampling
2. Purposive Sampling
3. Quota Sampling
4. Snowball Sampling
5. Voluntary Response Sampling

---

# 10. Probability Sampling

Probability sampling is particularly important when the goal is to make statistically defensible inferences about a population.

---

## 10.1 Simple Random Sampling

In **Simple Random Sampling**, every population member has an equal chance of being selected under the sampling design.

### Example

Suppose a university has 10,000 students and we need a sample of 500.

We can assign every student an ID:

```text
Student 1
Student 2
Student 3
...
Student 10,000
```

Then randomly select 500 IDs.

### Advantages

* Simple concept
* Reduces selection bias
* Easy to understand
* Suitable when a complete sampling frame is available

### Limitations

* Requires a reliable sampling frame
* Can be difficult for geographically dispersed populations
* May not adequately represent small subgroups by chance

---

# 11. Systematic Sampling

In **Systematic Sampling**, members are selected at a regular interval from an ordered sampling frame.

A common sampling interval is:

[
k = \frac{N}{n}
]

where:

* (N) = population size
* (n) = desired sample size
* (k) = sampling interval

### Example

Suppose:

[
N = 10,000
]

and:

[
n = 500
]

Then:

[
k = \frac{10,000}{500} = 20
]

After choosing an appropriate random starting position, we select every 20th individual.

```text
Random Start
     ↓
7 → 27 → 47 → 67 → 87 → 107 → ...
```

### Advantages

* Simple to implement
* Efficient for ordered lists
* Often easier than simple random sampling

### Limitation

Systematic sampling can be problematic if the ordering contains a hidden **periodic pattern** that aligns with the sampling interval.

---

# 12. Stratified Sampling

In **Stratified Sampling**, the population is divided into meaningful subgroups called **strata**, and samples are selected from each stratum.

### Example

Suppose a university population consists of:

* 60% undergraduate students
* 30% master's students
* 10% PhD students

We can divide the population into these three strata and sample from each.

```text
                Population
                    |
       ┌────────────┼────────────┐
       ↓            ↓            ↓
 Undergraduate   Master's       PhD
       |            |            |
       ↓            ↓            ↓
    Sample        Sample        Sample
```

### Why Use Stratified Sampling?

It is useful when important subgroups should be adequately represented.

### Advantages

* Ensures representation of important groups
* Can improve precision
* Allows subgroup analysis
* Useful when populations are heterogeneous but strata are internally similar

### Common Allocation Approaches

#### Proportionate Allocation

The sample size from each stratum is proportional to its population size.

[
n_h = n \frac{N_h}{N}
]

where:

* (n_h) = sample size for stratum (h)
* (n) = total sample size
* (N_h) = population size of stratum (h)
* (N) = total population size

#### Disproportionate Allocation

Different numbers of observations may be selected from different strata when some groups need additional representation.

---

# 13. Cluster Sampling

In **Cluster Sampling**, the population is divided into naturally occurring groups called **clusters**.

Instead of selecting individuals from the entire population, researchers select clusters and then survey members within the selected clusters according to the design.

### Example

Suppose we want to survey students across a large country.

Instead of randomly selecting students from the entire country, we could:

```text
Country
   |
   ├── University A
   ├── University B
   ├── University C
   ├── University D
   └── University E
          |
          ↓
    Select Universities
          |
          ↓
    Survey Students
```

### Advantages

* Lower travel and operational costs
* Useful for geographically distributed populations
* Does not always require a complete list of individuals

### Limitation

Individuals within the same cluster may be similar to one another. This can reduce the amount of independent information gained from each additional observation.

---

# 14. Multistage Sampling

**Multistage Sampling** involves selecting samples through multiple stages.

### Example

A national education survey might use:

```text
Country
   ↓
Provinces
   ↓
Districts
   ↓
Schools
   ↓
Classes
   ↓
Students
```

At each stage, a sampling method is applied.

### Advantages

* Highly practical for large-scale surveys
* Reduces logistical complexity
* Useful for geographically distributed populations

### Limitation

The design can become statistically and operationally complex, and estimates often need to account for the multistage design.

---

# 15. Non-Probability Sampling

Non-probability sampling is often used when probability sampling is difficult, expensive, or unnecessary for the research objective.

However, it generally provides weaker foundations for population-wide statistical inference because selection probabilities are not fully known.

---

# 16. Convenience Sampling

In **Convenience Sampling**, researchers select individuals who are easiest to access.

### Example

A student wants to conduct a survey and asks classmates because they are readily available.

### Advantages

* Very easy
* Fast
* Inexpensive
* Useful for exploratory research

### Limitations

* High risk of selection bias
* May not represent the population
* Weak basis for generalization

> Convenience samples are easy to collect but can be difficult to defend when making claims about a broader population.

---

# 17. Purposive Sampling

In **Purposive Sampling**, researchers intentionally select participants who meet specific criteria or possess relevant characteristics.

### Example

If researchers are studying professional data scientists, they may specifically recruit people with substantial industry experience in data science.

### Advantages

* Useful for specialized populations
* Helpful in qualitative research
* Allows targeted participant selection

### Limitations

* Subjective selection
* Potential selection bias
* Limited generalizability

---

# 18. Quota Sampling

In **Quota Sampling**, researchers establish target numbers for different groups and then select participants until those quotas are filled.

### Example

A survey may require:

| Group   | Required Participants |
| ------- | --------------------: |
| Group A |                   100 |
| Group B |                   100 |
| Group C |                    50 |

Researchers continue recruiting until each quota is reached.

### Difference from Stratified Sampling

The two methods may look similar, but the selection process differs:

* **Stratified sampling:** selection within strata uses a probability-based mechanism.
* **Quota sampling:** participants within groups are typically selected non-randomly.

---

# 19. Snowball Sampling

**Snowball Sampling** is useful when the target population is difficult to identify or reach.

The process typically works as:

```text
Initial Participants
        ↓
Refer Other Participants
        ↓
New Participants
        ↓
Further Referrals
        ↓
Larger Sample
```

### Example

A researcher interviews a specialized professional group and asks participants to recommend other qualified individuals.

### Advantages

* Useful for hard-to-reach populations
* Can help researchers build a sample through social connections

### Limitations

* Participants may recruit people similar to themselves
* Strong risk of network-related selection bias
* Limited generalizability

---

# 20. Voluntary Response Sampling

In **Voluntary Response Sampling**, people choose whether to participate.

### Example

A website posts:

> "Click here to participate in our survey."

People who are particularly interested in the topic may be more likely to respond.

This can produce **voluntary response bias**.

---

# 21. Sampling Bias

**Sampling bias** occurs when the sampling process systematically favors some members of the population over others, causing the sample to differ in important ways from the target population.

### Example

Suppose a company wants to measure customer satisfaction but only surveys customers who recently made a purchase.

Customers who stopped using the service are excluded, so the sample may provide an overly positive picture.

---

# 22. Common Sources of Sampling Bias

## 22.1 Undercoverage

Some parts of the target population are inadequately represented or excluded from the sampling frame.

## 22.2 Nonresponse Bias

Selected individuals do not respond, and responders differ systematically from nonresponders.

## 22.3 Voluntary Response Bias

People choose whether to participate, and those who participate may have stronger opinions.

## 22.4 Selection Bias

The sampling procedure systematically favors certain members.

## 22.5 Survivorship Bias

Analysis focuses only on entities that remain visible or successful while ignoring those that disappeared or failed.

---

# 23. Sampling Error

**Sampling error** is the difference between a population quantity and the corresponding value estimated from a sample due to observing only part of the population.

For example:

[
\text{Sampling Error}
=====================

## \text{Sample Statistic}

\text{Population Parameter}
]

Suppose:

[
\text{True Population Mean} = 70
]

and:

[
\text{Sample Mean} = 72
]

Then:

[
\text{Sampling Error} = 72 - 70 = 2
]

Sampling error is not necessarily evidence of a mistake. It is a natural consequence of sampling.

---

# 24. Sampling Error vs Sampling Bias

These concepts should not be confused.

| Sampling Error                                    | Sampling Bias                                              |
| ------------------------------------------------- | ---------------------------------------------------------- |
| Random difference between sample and population   | Systematic difference caused by selection process          |
| Usually decreases with appropriate larger samples | Increasing sample size does not automatically remove it    |
| Natural consequence of sampling                   | Problem with the sampling design                           |
| Can be quantified under probability sampling      | Often difficult to quantify without additional information |

### Key Idea

> **A larger biased sample can still be biased.**

For example, surveying 100,000 people from only one demographic group does not necessarily represent the entire population.

---

# 25. Sample Size

Sample size refers to the number of observations included in the sample.

Choosing an appropriate sample size is important because:

* Too small a sample may produce unstable estimates.
* An unnecessarily large sample wastes time and resources.

Sample size depends on factors such as:

* Population size
* Desired precision
* Confidence level
* Variability in the population
* Expected effect size
* Sampling design
* Available resources
* Study objectives

---

# 26. Confidence Level and Margin of Error

For many survey estimates, researchers consider the **confidence level** and **margin of error**.

### Confidence Level

A commonly used confidence level is:

[
95%
]

Other common choices include:

[
90%, \quad 95%, \quad 99%
]

### Margin of Error

The margin of error represents a range around an estimate that reflects sampling uncertainty under the assumptions of the statistical method.

For a population proportion under a simple random sampling approximation:

[
ME =
z_{\alpha/2}
\sqrt{
\frac{p(1-p)}{n}
}
]

where:

* (ME) = margin of error
* (z_{\alpha/2}) = critical value
* (p) = estimated population proportion
* (n) = sample size

For a 95% confidence level:

[
z_{\alpha/2} \approx 1.96
]

under the standard normal approximation.

---

# 27. Relationship Between Sample Size and Precision

Generally:

[
\text{Larger Sample}
\Rightarrow
\text{Smaller Sampling Uncertainty}
]

However, the improvement is subject to **diminishing returns**.

For example, increasing the sample from 100 to 1,000 can substantially improve precision, but increasing it from 100,000 to 101,000 usually provides a much smaller improvement.

The important point is that **sample quality matters alongside sample size**.

---

# 28. Sampling Distribution

A **sampling distribution** is the probability distribution of a statistic obtained from repeated samples of the same population under a specified sampling procedure.

For example, imagine repeatedly selecting samples of size (n) and calculating the sample mean each time:

```text
Population
    |
    ├── Sample 1 → Mean 1
    ├── Sample 2 → Mean 2
    ├── Sample 3 → Mean 3
    ├── Sample 4 → Mean 4
    ├── Sample 5 → Mean 5
    └── ...
             ↓
     Sampling Distribution
        of the Mean
```

Sampling distributions are fundamental to:

* Confidence intervals
* Hypothesis testing
* Standard errors
* Statistical inference
* The Central Limit Theorem

---

# 29. Standard Error

The **standard error (SE)** measures the variability of a sample statistic across repeated samples under a specified sampling design and statistical model.

For the sample mean under simple random sampling from a population with standard deviation (\sigma):

[
SE(\bar{x}) =
\frac{\sigma}{\sqrt{n}}
]

If the population standard deviation is unknown, it is commonly estimated using the sample standard deviation (s):

[
SE(\bar{x}) =
\frac{s}{\sqrt{n}}
]

This illustrates an important relationship:

[
n \uparrow
\quad \Rightarrow \quad
SE \downarrow
]

---

# 30. Central Limit Theorem and Sampling

The **Central Limit Theorem (CLT)** is closely related to sampling distributions.

Under appropriate conditions, as sample size increases, the sampling distribution of the sample mean tends toward a normal distribution, even when the original population is not normally distributed.

Conceptually:

```text
Population Distribution
        ↓
Repeated Sampling
        ↓
Sample Means
        ↓
Sampling Distribution
        ↓
Approximately Normal
under appropriate conditions
```

The CLT is one of the foundations that makes many statistical inference procedures possible.

In machine learning, class imbalance can create additional sampling challenges.

Suppose a dataset contains:

| ------- | -----------: |
| Class A |       95,000 |
| Class B |        5,000 |

The minority class represents only:

[
5%
]

Randomly sampling a small subset may further reduce minority-class representation.

Possible approaches include:

* Stratified sampling
* Oversampling
* Undersampling
* Class weighting

The appropriate method depends on the modeling objective and evaluation strategy.

# 33. Stratified Sampling in Machine Learning

When creating training and testing datasets, **stratified sampling** can be particularly useful for classification problems.

For example, if the original dataset contains:

```text
Class A = 80%
Class B = 20%
```

a stratified split attempts to preserve approximately the same class proportions in the training and testing sets.

```text
Original Dataset
A = 80%
B = 20%
       |
       ↓
Stratified Split
       |
   ┌───┴────┐
   ↓        ↓
Training   Testing
 A ≈ 80%   A ≈ 80%
 B ≈ 20%   B ≈ 20%
```

This can help prevent accidental changes in class distribution caused by random splitting.

---

# 34. Sampling With Replacement

In **sampling with replacement**, an observation can be selected more than once.

After an observation is selected, it is returned to the population before the next selection.

```text
Population
A B C D E

Select C
↓
Return C

A B C D E

Select C again
```

Therefore, the same observation may appear multiple times in the resulting sample.

This idea is particularly important in **bootstrap sampling**.

---

# 35. Sampling Without Replacement

In **sampling without replacement**, once an observation has been selected, it cannot be selected again.

```text
Population
A B C D E

Select C

Remaining
A B D E
```

Many traditional survey sampling procedures use sampling without replacement.

---

# 36. Bootstrap Sampling

**Bootstrapping** is a resampling technique in which repeated samples are generated from the observed sample, usually with replacement.

Suppose the original sample is:

[
[10, 20, 30, 40, 50]
]

A bootstrap sample of the same size might be:

[
[20, 20, 40, 50, 10]
]

Another could be:

[
[30, 30, 10, 50, 50]
]

Many bootstrap samples can be generated to study the variability of a statistic.

Bootstrap methods are widely used for:

* Estimating uncertainty
* Confidence intervals
* Standard errors
* Model evaluation
* Statistical inference

# 38. Sampling Frame

A **sampling frame** is the practical list or mechanism from which the sample is selected.

Examples:

* Student enrollment database
* Customer database

A good sampling frame should:

* Minimize duplicate entries

# 42. Common Sampling Mistakes

### Mistake 1: Assuming a Large Sample Is Automatically Representative

A large sample can still be biased.

### Mistake 2: Ignoring Nonresponse

If many selected participants do not respond, the final sample may differ from the original sample.

### Mistake 3: Using Convenience Sampling Without Acknowledging Its Limitations

Convenience samples are not automatically representative.

### Mistake 4: Ignoring Population Subgroups

Important minority groups may be poorly represented in a simple random sample.

### Mistake 5: Using an Inappropriate Sampling Frame

If the sampling frame does not adequately cover the target population, selection problems can occur.

### Mistake 6: Confusing Sampling Error With Bias

Increasing sample size primarily reduces random sampling variability; it does not automatically correct systematic bias.

# 44. Sampling and Generalization

One of the major goals of sampling is to make **inferences about a population from sample data**.

```text
Population
    ↓
Sampling
    ↓
Sample
    ↓
Statistical Analysis
    ↓
Estimate / Inference
    ↓
Population Conclusion
```

However, generalization should only be made when the sampling design and study conditions support it.

The broader the claim, the more important representativeness becomes.

# 48. Key Takeaways

> **Sampling allows us to learn about a large population by studying a smaller group.**

The most important concepts to remember are:

1. **Population** is the complete group of interest.
2. **Sample** is a subset of the population.
3. **Sampling** is the process of selecting that subset.
4. **Probability sampling** uses a defined probability-based selection mechanism.
5. **Non-probability sampling** uses non-random or partly non-random selection.
6. **Simple random sampling** gives each population member an equal selection chance under the design.
7. **Stratified sampling** divides the population into important subgroups.
8. **Cluster sampling** selects naturally occurring groups.
9. **Systematic sampling** selects units at regular intervals.
10. **Multistage sampling** performs sampling through multiple stages.
11. **Convenience sampling** selects easily accessible participants.
12. **Sampling bias** is a systematic problem caused by the selection process.
13. **Sampling error** is random variation resulting from observing a sample rather than the entire population.
14. **Larger samples generally reduce sampling variability**, but they do not automatically eliminate bias.
15. **A representative sample is more important than simply having a large sample.**
16. Sampling is a fundamental concept in **statistics, data science, machine learning, surveys, and scientific research**.

# 49. Final Concept

The central idea behind sampling can be expressed as:

[
\boxed{
\text{Well-Designed Sample}
\rightarrow
\text{Reliable Statistical Information}
\rightarrow
\text{Better Population Inference}
}

The goal of sampling is **not simply to collect fewer observations**.

The real goal is to collect **the right observations, using an appropriate method, so that the sample provides reliable information about the population**.
