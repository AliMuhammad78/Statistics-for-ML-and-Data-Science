# T-Tests

## 1. Introduction

A **t-test** is a statistical hypothesis test used to determine whether there is sufficient evidence to conclude that a population mean, or the difference between means, is statistically significant.

T-tests are especially useful when:

* The sample size is relatively small.
* The population standard deviation is unknown.
* The data are approximately normally distributed, particularly for small samples.
* We want to compare means.

The fundamental idea is to compare an observed difference against the amount of variability we would expect from sampling.

---

# 2. Why Do We Need a T-Test?

Suppose a teacher wants to know whether the average score of students is different from **70**.

A sample of 15 students gives:

[
\bar{x} = 74
]

The question is:

> Is the difference between 74 and 70 large enough to be considered statistically significant, or could it reasonably occur because of random sampling variation?

A t-test helps answer this question.

It does not simply ask whether two numbers are different. It evaluates whether the observed difference is large **relative to the uncertainty in the estimate**.

---

# 3. The Basic Idea

A t-test is based on a ratio:

[
t =
\frac{\text{Observed Difference}}
{\text{Estimated Standard Error}}
]

In general:

[
\boxed{
t =
\frac{\text{Estimate} - \text{Null Value}}
{\text{Standard Error}}
}
]

A large absolute value of (t) means that the observed result is far from what would be expected under the null hypothesis.

A value of (t) close to zero means that the observed difference is relatively small compared with its standard error.

---

# 4. Why Is It Called the "Student's t-Test"?

The name has an interesting history.

The t-distribution and t-test were developed by **William Sealy Gosset**, an English statistician who worked for the Guinness brewery.

Guinness had restrictions on employees publishing under their own names. Therefore, Gosset published his statistical work in **1908** using the pseudonym:

> **Student**

His paper was titled:

> *The Probable Error of a Mean*

Because Gosset published under the name **Student**, the resulting distribution became known as the **Student's t-distribution**, and the associated procedure became known as the **Student's t-test**.

### Important

"Student" does **not** mean that the test is only for students.

It refers to **William Sealy Gosset's publishing pseudonym, "Student."**

---

# 5. The Student's t-Distribution

The t-test uses the **Student's t-distribution**.

The t-distribution resembles the standard normal distribution but has **heavier tails**.

```text
Standard Normal
       /\
      /  \
     /    \
____/      \____

Student's t
      /\
     /  \
____/    \____
  heavier tails
```

The heavier tails reflect additional uncertainty that arises when the population standard deviation is unknown and must be estimated from the sample.

---

# 6. Why Not Always Use the Z-Test?

When the population standard deviation (\sigma) is known, a z-test can be used in the appropriate setting.

However, in real-world situations, the population standard deviation is often unknown.

Instead, we estimate it using the sample standard deviation (s).

This introduces additional uncertainty.

The t-distribution accounts for this uncertainty.

### General Rule

| Situation                                               | Common Test                      |
| ------------------------------------------------------- | -------------------------------- |
| Population standard deviation known                     | Z-test                           |
| Population standard deviation unknown                   | T-test                           |
| Small sample with unknown population standard deviation | T-test is particularly important |

The choice of test also depends on the study design and assumptions, not only on sample size.

---

# 7. Main Types of T-Tests

There are three major forms of t-tests:

```text
                       T-TESTS
                          |
          ┌───────────────┼────────────────┐
          |               |                |
   One-Sample T-Test  Independent     Paired T-Test
                       T-Test
```

1. **One-Sample T-Test**
2. **Independent Two-Sample T-Test**
3. **Paired-Samples T-Test**

---

# 8. One-Sample T-Test

A **one-sample t-test** compares the mean of a single sample with a known or hypothesized population value.

### Example

Suppose a company claims:

[
\mu = 50
]

We collect a sample of 20 products and obtain:

[
\bar{x} = 53
]

We can use a one-sample t-test to determine whether there is sufficient evidence that the population mean differs from 50.

### Formula

[
t =
\frac{\bar{x}-\mu_0}
{s/\sqrt{n}}
]

where:

* (\bar{x}) = sample mean
* (\mu_0) = hypothesized population mean
* (s) = sample standard deviation
* (n) = sample size

### Example Question

> Is the average delivery time significantly different from 30 minutes?

---

# 9. Independent Two-Sample T-Test

An **independent two-sample t-test** compares the means of **two independent groups**.

### Example

Suppose we want to compare the average exam scores of:

* Group A
* Group B

The observations in Group A are independent of the observations in Group B.

The test asks:

[
H_0:\mu_1=\mu_2
]

versus an alternative such as:

[
H_A:\mu_1\neq\mu_2
]

### Example Question

> Is the average exam score of Class A significantly different from Class B?

---

# 10. Independent vs Welch's T-Test

There are two commonly encountered versions of the independent two-sample t-test.

### Student's Independent T-Test

The classical pooled version assumes equal population variances.

### Welch's T-Test

Welch's t-test does **not** require the two population variances to be equal.

In practical data analysis, **Welch's t-test is often preferred as the default two-sample t-test** when there is no strong reason to assume equal variances.

### Comparison

| Test                               | Equal Variance Assumption |
| ---------------------------------- | ------------------------- |
| Student's pooled two-sample t-test | Yes                       |
| Welch's t-test                     | No                        |

This distinction is important because "Student's t-test" can refer broadly to t-testing based on the Student t-distribution, while the specific independent-samples procedure may use either pooled Student's t-test or Welch's t-test.

---

# 11. Paired-Samples T-Test

A **paired t-test** is used when two measurements are naturally related or matched.

Instead of treating the two groups as independent, we analyze the **difference within each pair**.

### Examples

* Before vs after treatment
* Weight before vs after a program
* Test score before vs after training
* Measurement using Method A vs Method B on the same subjects

Suppose we have:

| Person | Before | After |
| ------ | -----: | ----: |
| 1      |     65 |    72 |
| 2      |     70 |    75 |
| 3      |     68 |    71 |
| 4      |     74 |    80 |

We calculate:

[
d_i = After_i - Before_i
]

Then perform a one-sample t-test on the differences.

### Important Idea

[
\boxed{
\text{Paired T-Test}
====================

\text{One-Sample T-Test on Differences}
}
]

---

# 12. Hypotheses in T-Tests

Every t-test involves hypotheses.

## Null Hypothesis

The **null hypothesis (H_0)** generally represents no difference or no effect.

For example:

[
H_0:\mu = \mu_0
]

or:

[
H_0:\mu_1=\mu_2
]

## Alternative Hypothesis

The **alternative hypothesis (H_A)** represents the effect or difference we are investigating.

For a two-sided test:

[
H_A:\mu\neq\mu_0
]

For a right-tailed test:

[
H_A:\mu>\mu_0
]

For a left-tailed test:

[
H_A:\mu<\mu_0
]

---

# 13. One-Tailed vs Two-Tailed T-Tests

T-tests can be conducted as **one-tailed** or **two-tailed** tests depending on the research hypothesis.

### Two-Tailed Test

Tests whether the parameter is different in either direction:

[
H_A:\mu\neq\mu_0
]

### Right-Tailed Test

Tests whether the parameter is greater:

[
H_A:\mu>\mu_0
]

### Left-Tailed Test

Tests whether the parameter is smaller:

[
H_A:\mu<\mu_0
]

The direction of the alternative hypothesis should be determined **before examining the results**, based on the research question.

---

# 14. Degrees of Freedom

The **degrees of freedom (df)** are an important part of a t-test because the shape of the t-distribution depends on them.

For a one-sample t-test:

[
df=n-1
]

For a paired t-test with (n) pairs:

[
df=n-1
]

For an independent two-sample t-test, the degrees of freedom depend on the variance assumption and the specific test.

For Welch's t-test, the degrees of freedom are estimated using the **Welch-Satterthwaite equation**.

---

# 15. Why Do Degrees of Freedom Matter?

As the degrees of freedom increase, the t-distribution becomes increasingly similar to the standard normal distribution.

Conceptually:

```text
Small df
   ↓
Heavier tails
   ↓
More uncertainty

Large df
   ↓
Less heavy tails
   ↓
Closer to Normal Distribution
```

As:

[
df \rightarrow \infty
]

the Student's t-distribution approaches the standard normal distribution.

---

# 16. P-Value in a T-Test

The **p-value** measures how compatible the observed result is with the null hypothesis, under the assumptions of the test.

A small p-value indicates that the observed result would be relatively unusual if the null hypothesis were true.

A commonly used significance level is:

[
\alpha=0.05
]

A simplified decision rule is:

[
p < \alpha
\Rightarrow
\text{Reject }H_0
]

and:

[
p \geq \alpha
\Rightarrow
\text{Fail to Reject }H_0
]

### Important

Failing to reject (H_0) does **not** prove that (H_0) is true.

It means there is insufficient evidence to reject it at the chosen significance level.

---

# 17. T-Statistic and P-Value

The relationship can be visualized conceptually:

```text
Observed Difference
        |
        ↓
   T-Statistic
        |
        ↓
   T-Distribution
        |
        ↓
     P-Value
        |
        ↓
Statistical Decision
```

A larger absolute t-statistic generally corresponds to a smaller two-sided p-value, assuming the degrees of freedom remain fixed.

---

# 18. Confidence Intervals and T-Tests

T-tests are closely connected to confidence intervals.

For a one-sample mean, a confidence interval can be written as:

[
\bar{x}
\pm
t_{\alpha/2,df}
\frac{s}{\sqrt{n}}
]

A two-sided hypothesis test and its corresponding confidence interval are linked.

For example, at the 5% significance level:

* If the null value lies **outside** the corresponding 95% confidence interval, the null hypothesis is rejected.
* If the null value lies **inside** the interval, the null hypothesis is not rejected.

---

# 19. Assumptions of T-Tests

The assumptions depend on the specific t-test and study design.

Common considerations include:

### 1. Independence

Observations should be independent where the test requires independence.

### 2. Continuous Outcome

The response variable is typically quantitative and approximately continuous.

### 3. Approximate Normality

For small samples, the relevant population distribution or paired differences should be reasonably close to normal.

### 4. Outliers

Extreme outliers can strongly influence means and t-test results.

### 5. Equal Variances

This is relevant to the classical pooled independent-samples t-test.

It is **not required for Welch's t-test**.

---

# 20. T-Test Workflow

A typical t-test analysis follows these steps:

```text
1. Define Research Question
          ↓
2. State H₀ and Hₐ
          ↓
3. Choose Appropriate T-Test
          ↓
4. Check Relevant Assumptions
          ↓
5. Calculate T-Statistic
          ↓
6. Calculate P-Value
          ↓
7. Compare P-Value with α
          ↓
8. Make Statistical Decision
          ↓
9. Report Effect Size + Confidence Interval
          ↓
10. Interpret in Context
```

# 21. How to Choose the Correct T-Test

A simple decision framework:

```text
                     What are you comparing?
                              |
                 ┌────────────┴────────────┐
                 |                         |
          One Sample                  Two Measurements/Groups
                 |                         |
                 ↓                         ↓
        Compare with a value       Are observations related?
                                           |
                              ┌────────────┴────────────┐
                              |                         |
                             Yes                        No
                              |                         |
                              ↓                         ↓
                       Paired T-Test          Independent T-Test
```

### Quick Guide

| Situation                                     | Appropriate Test              |
| --------------------------------------------- | ----------------------------- |
| One sample vs known value                     | One-sample t-test             |
| Two independent groups                        | Independent two-sample t-test |
| Two related measurements                      | Paired t-test                 |
| Two independent groups with unequal variances | Welch's t-test                |

# 22. Effect Size

Statistical significance does not tell us how large or practically important a difference is.

Therefore, it is useful to report an **effect size** alongside the p-value.

One commonly used effect size is **Cohen's (d)**.

For two independent groups, a simplified form is:

[
d =
\frac{\bar{x}_1-\bar{x}*2}{s*{\text{pooled}}}
]

A large sample can produce a statistically significant result even when the practical difference is small.

Therefore:

[
\boxed{
\text{Statistical Significance}
\neq
\text{Practical Significance}
}
]

# 23. T-Test vs Z-Test

| Feature                       | T-Test              | Z-Test                 |
| ----------------------------- | ------------------- | ---------------------- |
| Distribution                  | Student's t         | Standard normal        |
| Population standard deviation | Usually unknown     | Typically known        |
| Uses sample SD                | Yes                 | Usually no             |
| Small samples                 | Particularly useful | Depends on assumptions |
| Degrees of freedom            | Yes                 | No                     |
| Tail behavior                 | Heavier tails       | Lighter tails          |

The t-test becomes increasingly similar to the z-test as the degrees of freedom become large.

# 24. T-Test in Python

Python provides several convenient ways to perform t-tests.

The most common approach is through SciPy:

```python
from scipy import stats
```

Examples include:

```python
stats.ttest_1samp(...)
```

for a one-sample t-test,

```python
stats.ttest_ind(...)
```

for an independent two-sample t-test,

and:

```python
stats.ttest_rel(...)
```

for a paired t-test.

For independent groups, Welch's t-test can be performed by setting:

```python
equal_var=False
```

in `stats.ttest_ind()`.

# 25. Important Interpretation

Suppose a t-test produces:

[
t=2.45
]

and:

[
p=0.02
]

with:

[
\alpha=0.05
]

Since:

[
0.02 < 0.05
]

we reject the null hypothesis at the 5% significance level.

A proper conclusion would be:

> There is statistically significant evidence against the null hypothesis at the 5% significance level.

Avoid saying:

> "The null hypothesis is definitely false."

or:

> "There is a 2% probability that the null hypothesis is true."

Those statements are not correct interpretations of a p-value.

---

# 26. Limitations of T-Tests

T-tests are powerful but not appropriate for every situation.

Potential issues include:

* Strongly non-normal data in small samples
* Severe outliers
* Dependence between observations when independence is assumed
* Incorrect handling of paired data
* Very unequal variances when using the pooled test
* Multiple comparisons without appropriate adjustment
* Treating statistical significance as practical importance

When assumptions are seriously violated, alternative methods may be more appropriate.

Examples include:

* Mann–Whitney U test
* Wilcoxon signed-rank test
* Permutation tests
* Bootstrap methods
* Regression-based approaches
