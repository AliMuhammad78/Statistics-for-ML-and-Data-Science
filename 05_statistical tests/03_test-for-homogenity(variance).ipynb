## 1. What is Homogeneity?

**Homogeneity of variance** means that different groups have approximately the **same variance**.

In simple words:

> The spread or variability of the data is similar across groups.

For example, suppose we compare the exam scores of three classes:

* Class A
* Class B
* Class C

If all three classes have similar variability in their scores, the data satisfy the **homogeneity of variance** assumption.

This assumption is important for several statistical methods, including:

* T-tests
* ANOVA
* Some regression and experimental procedures

---

## 2. Why Do We Check Homogeneity?

Before applying certain statistical tests, we need to check whether the groups have similar variances.

If the assumption is seriously violated, the usual test may not perform as expected.

Two common tests for checking equality of variances are:

```text
          Homogeneity of Variance
                    |
          ┌─────────┴─────────┐
          ↓                   ↓
   Levene's Test        Bartlett's Test
```

---

# 3. Levene's Test

**Levene's test** checks whether the variances of two or more groups are equal.

### Hypotheses

[
H_0:
\sigma_1^2 = \sigma_2^2 = \cdots = \sigma_k^2
]

The null hypothesis states that the group variances are equal.

[
H_A:
\text{At least one variance is different}
]

### Interpretation

* **p-value > 0.05:** No significant evidence that the variances differ.
* **p-value < 0.05:** Evidence that the variances are different.

### Key Advantage

Levene's test is relatively **robust to departures from normality**, making it a common practical choice.

---

# 4. Bartlett's Test

**Bartlett's test** also checks whether multiple groups have equal variances.

### Hypotheses

[
H_0:
\sigma_1^2 = \sigma_2^2 = \cdots = \sigma_k^2
]

[
H_A:
\text{At least one variance is different}
]

### Interpretation

* **p-value > 0.05:** No significant evidence of unequal variances.
* **p-value < 0.05:** Evidence of unequal variances.

### Important Difference

Bartlett's test is **more sensitive to departures from normality**.

Therefore, when the data are not reasonably normally distributed, **Levene's test is often preferred**.

---

# 5. Levene's vs Bartlett's Test

| Feature                 | Levene's Test | Bartlett's Test              |
| ----------------------- | ------------- | ---------------------------- |
| Tests equal variances   | Yes           | Yes                          |
| Robust to non-normality | More robust   | Less robust                  |
| Normality assumption    | Less strict   | More important               |
| Common practical choice | Yes           | When normality is reasonable |

---

# 6. Key Takeaway

> **Both Levene's and Bartlett's tests check homogeneity of variance, but Levene's test is generally more robust when the data may not be normally distributed.**

A simple rule to remember:

[
\boxed{
\text{Homogeneity of Variance}
\rightarrow
\begin{cases}
\text{Levene's Test} & \text{More robust}\
\text{Bartlett's Test} & \text{More sensitive to non-normality}
\end{cases}
}
]

# 1. Levene Test

````python
import pandas as pd
import numpy as np 
import seaborn as sns 
from scipy import stats 
````

````python
group1 = [1.3 , 4.5 , 9.2 , 8.4 , 1.9 , 2.2 ]
group2 = [2.3 , 9.5 , 3.2 , 8.0 , 1.1 , 2.2 ]

t_test , p = stats.levene(group1 , group2)

# print("T-statistcs : " , {t_statistics} )
# print("P-value :" , p)



if p > 0.05 : 
    print(f"p-value : {p}  , group1 variance is equal to the group2 variance (failed to reject H0)" )
else :
    print(f"p-value : {p}  , group1 variance is not equal to the group2 variance (reject H0)" )
````

# 2. Barlett's Test

````python
group1 = [1.3 , 4.5 , 9.2 , 8.4 , 1.9 , 2.2 ]
group2 = [2.3 , 9.5 , 3.2 , 8.0 , 1.1 , 2.2 ]

t_test , p = stats.bartlett(group1 , group2)

# print("T-statistcs : " , {t_statistics} )
# print("P-value :" , p)



if p > 0.05 : 
    print(f"p-value : {p}  , group1 variance is equal to the group2 variance (failed to reject H0)" )
else :
    print(f"p-value : {p}  , group1 variance is not equal to the group2 variance (reject H0)" )
````
