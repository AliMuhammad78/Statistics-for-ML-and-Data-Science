# Measures of Central Tendency

Measures of central tendency describe the **typical or central value** of a dataset.

| Measure | Best Used For | Main Advantage | Main Limitation |
|---|---|---|---|
| **Arithmetic Mean** | Symmetric numerical data | Uses all observations; easy to calculate | Sensitive to outliers |
| **Geometric Mean** | Growth rates, ratios, percentages | Suitable for multiplicative data | Cannot handle zero/negative values in the usual form |
| **Harmonic Mean** | Rates, ratios, speeds | Best for averaging rates | Sensitive to very small values; cannot handle zero |
| **Median** | Skewed data, ordinal data | Resistant to outliers | Does not use the exact magnitude of all values |
| **Mode** | Categorical or discrete data | Only measure usable for nominal data; simple | May have multiple modes or no unique mode |

---

## 1. Arithmetic Mean

\[
\bar{x} = \frac{\sum x_i}{n}
\]

**Merits**
- Uses **all values** in the dataset.
- Easy to calculate and understand.
- Useful for further statistical analysis.
- Most commonly used measure of center.

**Demerits**
- Strongly affected by **outliers**.
- Not ideal for highly skewed data.
- Cannot normally be used directly with categorical data.

---

## 2. Geometric Mean

\[
GM = \sqrt[n]{x_1x_2...x_n}
\]

**Merits**
- Useful for **growth rates, returns, ratios, and percentages**.
- Less affected by extremely large values than the arithmetic mean.
- Appropriate for **multiplicative changes**.

**Demerits**
- More difficult to calculate and interpret.
- Requires positive values in its standard form.
- Not suitable for data containing zero or negative values.

---

## 3. Harmonic Mean

\[
HM = \frac{n}{\sum \frac{1}{x_i}}
\]

**Merits**
- Best suited for **rates and ratios**.
- Useful for averaging quantities such as **speed, price per unit, or rates**.
- Gives greater importance to smaller values.

**Demerits**
- Highly affected by very small values.
- Cannot be calculated when a value is zero.
- Less intuitive and less commonly used than the arithmetic mean.

---

## 4. Median

The **middle value** after arranging the data in ascending or descending order.

**Merits**
- Resistant to **outliers**.
- Excellent for **skewed distributions**.
- Can be used with **ordinal data**.
- Easy to understand.

**Demerits**
- Does not use the exact value of every observation.
- Less suitable for some advanced mathematical/statistical calculations.
- Requires sorting the data.

---

## 5. Mode

The value that occurs **most frequently**.

**Merits**
- Very simple to understand.
- Can be used with **categorical/nominal data**.
- Not affected by extreme values.
- Useful when the **most common category/value** is important.

**Demerits**
- A dataset may have **no mode**.
- A dataset may have **more than one mode**.
- Does not use all observations.

---

## Quick Selection Guide

| Data Situation | Preferred Measure |
|---|---|
| Normal/symmetric numerical data | **Arithmetic Mean** |
| Skewed data or outliers | **Median** |
| Categorical data | **Mode** |
| Growth rates / compound changes | **Geometric Mean** |
| Rates / speeds / ratios | **Harmonic Mean** |

> **Mean → overall numerical average**  
> **Median → middle value**  
> **Mode → most frequent value**  
> **Geometric Mean → multiplicative change**  
> **Harmonic Mean → rates and ratios**
