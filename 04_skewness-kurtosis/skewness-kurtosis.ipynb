#  Understanding Distribution Shape: Skewness & Kurtosis

When analyzing data, simply looking at the "Mean" or "Standard Deviation" isn't enough. We need to understand the **shape** of the distribution to choose the right models and transformations.

Two key metrics describe this shape: **Skewness** (Symmetry) and **Kurtosis** (Tailedness).

---

## **1. Skewness (Measure of Asymmetry)**

Skewness measures the lack of symmetry in data distribution. It tells us if data is concentrated on one side.

### **Types of Skewness:**

* **Normal Distribution (Symmetrical):**
* **Skewness ≈ 0**
* The Left side is a mirror image of the Right side.
* **Key Rule:** Mean = Median = Mode.

## **2. Kurtosis (Measure of Tailedness)**

Kurtosis measures the "heaviness" of the tails (outliers) relative to a normal distribution. It describes how extreme the data points are.
*(Note: In Python's `scipy`, we usually use "Excess Kurtosis", where Normal = 0).*

### **Types of Kurtosis:**

#### **Mesokurtic (Normal-like):**
* **Kurtosis ≈ 0** (Excess) or 3 (Fisher).
* Standard Normal Distribution.
* Moderate tails, moderate peak.


#### **Leptokurtic (Heavy Tails):**
* **Kurtosis > 0**
* **Shape:** Sharp, tall peak and **fat/heavy tails**.
* **Meaning:** The data has **many outliers**. There is a high risk of extreme events.
* *Mnemonic:* "Lepto" sounds like "Leap" (Leaping high).


#### **Platykurtic (Light Tails):**
* **Kurtosis < 0**
* **Shape:** Flat peak and **thin/light tails**.
* **Meaning:** The data has **few outliers**. Values are spread out more evenly.
* *Mnemonic:* "Platy" sounds like "Plate" (Flat).

#  Role of Skewness & Kurtosis in Data Analysis

Skewness and kurtosis help us understand important characteristics of a dataset and guide better analysis and decision-making.

## 🔹 Role of Skewness
- Helps identify **asymmetry** in the data
- Indicates whether the **mean or median** is more representative
- Helps detect potential **outliers**
- Guides the choice of **data transformation**
- Helps assess assumptions for **statistical and ML models**

## 🔹 Role of Kurtosis
- Helps identify **extreme values and outliers**
- Shows the **tail behavior** of the data
- Helps assess the impact of unusual observations
- Supports **data quality and distribution analysis**
- Helps check assumptions in **statistical modeling**

 

### **Summary Table**

| Metric | Value | Interpretation | Action Required? |
| --- | --- | --- | --- |
| **Skewness** | `Between -0.5 and 0.5` | Fairly Symmetrical | No |
|  | `Between -1 and -0.5` OR `0.5 and 1` | Moderately Skewed | Maybe (Square Root) |
|  | `Less than -1` OR `Greater than 1` | Highly Skewed | **Yes** (Log / Box-Cox) |
| **Kurtosis** | `High (> 1)` | Heavy Outliers | Check for outliers! |

## Practice on titanic 

```python
import pandas as pd
import numpy as np
import seaborn as sns
from scipy import stats


df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
stats.kurtosis(df["age"] , nan_policy="omit"  , fisher= False)
```

```python
stats.kurtosis(df["fare"] , nan_policy="omit"  , fisher= False)
```

```python
stats.skew(df["fare"])
```

```python
sns.histplot(df , x="fare" , kde=True)
```

```python
import numpy as np 
from scipy.stats import skew , kurtosis


np.random.seed(0)
data = np.random.normal( 0 , 1 , 100)


data_skewness = skew(data)
data_kurtosis = kurtosis(data , fisher=True)
 

print(f"Skewness is {data_skewness}")
print(f"Kurtosis is {data_kurtosis}")
```

```python
sns.histplot(data , kde=True , bins=20)
```

#### Let us take the data of third class people only and see the kurtosis 

```python
df_third = df[ df["class"] =="Third" ]
stats.kurtosis(df_third["fare"] , fisher=True)
```

```python
stats.skew(df_third["fare"])
```

```python
sns.histplot(df_third , x="fare"  , kde=True , bins=30, hue="class")
```

#### Let us take the data of second class people only and see the kurtosis 

```python
df_second = df[ df["class"] =="Second" ]
print("The kurtosis of second class is " , stats.kurtosis(df_second["fare"] , fisher=True))
print("The skewness of second class is  "  , stats.skew(df_second["fare"] ))
sns.histplot(df_second , x="fare" ,kde=True )
```

#### Let us take the data of first class people only and see the kurtosis 

```python
df_first = df[df["class"]=="First"]
print("The kurtosis of first class is " , stats.kurtosis(df_first["fare"] , fisher=True))
print("The skewness of first class is  "  , stats.skew(df_first["fare"] ))
sns.histplot(df_first , x="fare" ,kde=True )
```
