# Distributions Commonly Used in Data Science and Machine Learning

Distributions are mathematical models that help us understand the **pattern, shape, variability, likelihood, and underlying behavior of data**.

In data science and machine learning, distributions are useful for:

- Understanding the shape and structure of data
- Identifying skewness, spread, outliers, and unusual patterns
- Modeling different types of variables
- Selecting appropriate statistical methods
- Performing hypothesis testing
- Estimating uncertainty
- Creating confidence intervals
- Choosing appropriate machine learning models and loss functions
- Generating synthetic data and simulations
- Detecting anomalies
- Modeling probabilities and predictions

---

# 1. Bernoulli Distribution

## What it represents

The Bernoulli distribution represents a variable with only **two possible outcomes**.

For example:

- Yes / No
- Success / Failure
- Fraud / Not Fraud
- Click / No Click
- Customer Churn / No Churn
- Disease / No Disease

The random variable can take:

$$
X \in \{0,1\}
$$

where:

$$
P(X=1)=p
$$

and:

$$
P(X=0)=1-p
$$

## Mean

$$
E[X]=p
$$

## Variance

$$
Var(X)=p(1-p)
$$

## Data Science Applications

- Binary classification
- Logistic regression
- Conversion prediction
- Click-through rate prediction
- Churn prediction
- Fraud detection

---

# 2. Binomial Distribution

## What it represents

The Binomial distribution represents the **number of successes in a fixed number of independent trials**.

For example:

- Number of customers who purchase out of 100 visitors
- Number of defective products in a batch
- Number of students passing an exam
- Number of successful email conversions

If:

- $n$ = number of trials
- $p$ = probability of success

Then:

$$
X \sim Binomial(n,p)
$$

## Mean

$$
E[X]=np
$$

## Variance

$$
Var(X)=np(1-p)
$$

## Data Science Applications

- A/B testing
- Conversion analysis
- Success/failure experiments
- Binary event modeling
- Statistical inference for proportions

---

# 3. Poisson Distribution

## What it represents

The Poisson distribution models the **number of events occurring within a fixed interval of time, area, distance, or space**.

Examples:

- Number of website visits per minute
- Number of customer arrivals per hour
- Number of accidents per day
- Number of support tickets received
- Number of network failures

$$
X \sim Poisson(\lambda)
$$

where $\lambda$ represents the average rate of occurrence.

## Mean and Variance

$$
E[X]=\lambda
$$

$$
Var(X)=\lambda
$$

## Data Science Applications

- Count data
- Event prediction
- Customer arrival modeling
- Traffic analysis
- Call center analysis
- Poisson regression

---

# 4. Geometric Distribution

## What it represents

The Geometric distribution models the **number of trials required until the first success**.

Examples:

- Number of calls until a customer responds
- Number of attempts until a system succeeds
- Number of emails sent before receiving a reply

## Data Science Applications

- Sequential event analysis
- Retry systems
- Survival-like problems
- Modeling repeated attempts

---

# 5. Negative Binomial Distribution

## What it represents

The Negative Binomial distribution models the number of trials or failures before achieving a specified number of successes.

It is also commonly used for **count data with high variability**.

A major reason for using it in data science is when:

$$
Variance > Mean
$$

This situation is called **overdispersion**.

The Poisson distribution assumes:

$$
Mean = Variance
$$

If this assumption is not reasonable, the Negative Binomial distribution can often provide a better model.

## Applications

- Number of insurance claims
- Number of hospital visits
- Customer activity counts
- Crime/event counts
- Overdispersed count data

---

# 6. Uniform Distribution

## What it represents

In a Uniform distribution, values within a specific range have equal probability.

For a continuous interval:

$$
X \sim Uniform(a,b)
$$

## Mean

$$
E[X]=\frac{a+b}{2}
$$

## Variance

$$
Var(X)=\frac{(b-a)^2}{12}
$$

## Data Science and Machine Learning Applications

- Random initialization
- Random sampling
- Simulation
- Monte Carlo methods
- Randomized algorithms

---

# 7. Normal Distribution

## What it represents

The Normal distribution is one of the most important distributions in statistics, data science, and machine learning.

It is symmetric and characterized by:

- Mean $\mu$
- Variance $\sigma^2$

$$
X \sim N(\mu,\sigma^2)
$$

The distribution is commonly associated with the familiar bell-shaped curve.

## Mean

$$
E[X]=\mu
$$

## Variance

$$
Var(X)=\sigma^2
$$

## Why it is important

Many statistical methods are based on the Normal distribution or are approximately related to it through the **Central Limit Theorem**.

## Applications

- Statistical inference
- Linear regression assumptions
- Confidence intervals
- Hypothesis testing
- Feature analysis
- Gaussian models
- Anomaly detection

---

# 8. Standard Normal Distribution

A Standard Normal distribution is a Normal distribution with:

$$
\mu=0
$$

and:

$$
\sigma=1
$$

It is commonly written as:

$$
Z \sim N(0,1)
$$

## Standardization

A value can be transformed into a Z-score:

$$
Z=\frac{X-\mu}{\sigma}
$$

## Applications

- Feature standardization
- Outlier detection
- Comparing variables on different scales
- Statistical testing

---

# 9. Log-Normal Distribution

## What it represents

A variable follows a Log-Normal distribution when its logarithm follows a Normal distribution.

If:

$$
\ln(X) \sim N(\mu,\sigma^2)
$$

then:

$$
X
$$

follows a Log-Normal distribution.

## Characteristics

- Only positive values
- Right-skewed
- Can have a long right tail

## Examples

- Income
- Wealth
- Stock prices in simplified models
- Company sizes
- Transaction amounts

## Data Science Applications

When data is highly right-skewed, a log transformation is often considered:

$$
X'=\log(X)
$$

This can reduce skewness and make relationships easier to model.

---

# 10. Exponential Distribution

## What it represents

The Exponential distribution models **waiting time between events**.

Examples:

- Time until the next customer arrives
- Time until a machine fails
- Time until the next network request
- Waiting time between events

It is closely related to the Poisson distribution.

A useful relationship is:

- Poisson distribution → number of events
- Exponential distribution → waiting time between events

## Parameter

$$
\lambda
$$

## Mean

$$
E[X]=\frac{1}{\lambda}
$$

## Variance

$$
Var(X)=\frac{1}{\lambda^2}
$$

## Applications

- Survival analysis
- Reliability analysis
- Waiting time modeling
- Event timing
- Queueing systems

---

# 11. Gamma Distribution

## What it represents

The Gamma distribution is a flexible continuous distribution used for positive, right-skewed data.

It can model waiting times until multiple events occur.

The Exponential distribution can be considered a special case of the Gamma distribution.

## Characteristics

- Positive values
- Often right-skewed
- Flexible shape

## Applications

- Insurance claims
- Waiting times
- Rainfall amounts
- Financial data
- Bayesian statistics

# 12. Beta Distribution

## What it represents

The Beta distribution models values between:

$$
0 \leq X \leq 1
$$

It is especially useful for modeling **probabilities and proportions**.

For example:

- Conversion rate
- Click-through rate
- Probability of success
- Percentage/proportion values

The shape is controlled by two parameters:

$$
\alpha,\beta
$$

## Data Science Applications

- Bayesian statistics
- A/B testing
- Probability estimation
- Modeling proportions
- Bayesian updating

A major relationship:

$$
Beta + Bernoulli/Binomial
$$

is widely used in Bayesian analysis.

A Chi-Square test can help determine whether the two variables are independent.

 

#  Project: Normalizing Skewed Data

### **1. The Goal**

In this project, we practice handling data that does **not** follow a Normal Distribution (Bell Curve). Many Machine Learning models (like Linear Regression) perform better when data is normally distributed.

We will:

1. **Generate** data that is naturally skewed (leaning to the right).
2. **Visualize** it to confirm it is not normal.
3. **Apply Transformations** (Log and Box-Cox) to fix the skew.
4. **Compare** the results using histograms.

---

### **2. Key Libraries Used**

* `numpy` (`np`): Used for numerical math and generating our random dataset.
* `seaborn` (`sns`): Used for creating beautiful, easy-to-read statistical plots.
* `matplotlib.pyplot` (`plt`): Used to manage the figure size and layout of our charts.
* `scipy.stats`: The scientific library containing the **Box-Cox** transformation function.

---

### **3. Understanding the Transformations**

#### **Method A: Log Transformation**

* **Code:** `np.log(data + 1)`
* **Why use it?** The logarithm function compresses large numbers significantly more than small numbers. This "pulls in" the long tail of a right-skewed distribution.
* **Why +1?** Logarithm of zero (`log(0)`) is mathematically impossible (undefined). We add 1 to every data point to ensure we never crash the code if our data contains a zero.

#### **Method B: Box-Cox Transformation**

* **Code:** `stats.boxcox(data)`
* **Why use it?** This is a "smart" transformation. It tests various power transformations (like square root, log, reciprocal, etc.) and calculates a value called **Lambda ()**.
* **Lambda ():** This represents the power that best normalizes the data.
* If , no change is needed.
* If , it is equivalent to a Log transform.
* If , it is equivalent to a Square Root transform.

````python
 
import numpy as np 

import seaborn as sns 

import matplotlib.pyplot as plt 

from scipy import stats 

 
# 2. GENERATING SKEWED DATA
# We set a "seed" so that every time you run this code, you get the exact same random numbers.
# This makes your practice consistent.
np.random.seed(42)

# Here we generate 1000 random numbers that follow an "exponential" distribution.
# Exponential distributions are naturally "right-skewed" (tail goes to the right).
# 'scale=2' spreads the data out, and 'size=1000' means we create 1000 data points.
original_data = np.random.exponential(scale=2, size=1000)

 # 3. SETTING UP THE PLOT AREA

# This creates a blank canvas (figure) that is 14 inches wide and 4 inches tall.
plt.figure(figsize=(14, 4))

# 4. VISUALIZING THE ORIGINAL DATA
# subplot(1, 3, 1) means: "In a grid of 1 row and 3 columns, draw this in position 1".
plt.subplot(1, 3, 1)

# histplot draws the histogram (bars).
# 'kde=True' adds the Kernel Density Estimate (the smooth line curve on top).
sns.histplot(original_data, kde=True, color='red')

# giving the chart a title so we know what it is.
plt.title("Step 1: Original Data (Right Skewed)")


# 5. METHOD A: LOG TRANSFORMATION
# We take the Natural Logarithm (ln) of every number in our data.
# We add +1 because log(0) is impossible (undefined). 
# If your data has zeros, adding 1 is a common safety trick.
log_data = np.log(original_data + 1)

# Now we verify if it worked by plotting it in the second position (1, 3, 2).
plt.subplot(1, 3, 2)
sns.histplot(log_data, kde=True, color='blue')
plt.title("Step 2: Log Transformed Data")

# 6. METHOD B: BOX-COX TRANSFORMATION
# stats.boxcox is a smart function. It tries many power transformations and picks the best one.
# It returns two things: 
#   1. The transformed data (boxcox_data)
#   2. The best 'lambda' value it found (best_lambda)
boxcox_data, best_lambda = stats.boxcox(original_data)

# Plotting the result in the third position (1, 3, 3).
plt.subplot(1, 3, 3)
sns.histplot(boxcox_data, kde=True, color='green')

# In the title, we print which lambda value the computer chose (formatted to 2 decimal places).
plt.title(f"Step 3: Box-Cox (Best Lambda = {best_lambda:.2f})")


# 7. FINAL DISPLAY
# tight_layout fixes overlapping text or labels so the plot looks neat.
plt.tight_layout()

# Finally, show the plot window.
plt.show()
````

#  Diagnosis: Why did the Log Transformation fail?

### **1. The Problem: "Over-Correction"**

You successfully fixed the Right Skew, but you fixed it **too much**.

* **Original Data:** Leaning Right (Positive Skew).
* **Log Transformation:** Now Leaning Left (Negative Skew).

The Logarithm function is a very **aggressive** compressor. It shrinks large numbers drastically.

* It turned your large values (30) into small values (~3.4).
* It stretched your small values (near 0) into negative numbers (if you used log of values < 1) or very small decimals.

Because your data wasn't *extremely* skewed to begin with (it only went up to ~30, not millions), the Log function smashed the right side too hard, pushing the peak all the way to the right and creating a new tail on the left.

### **2. The Hierarchy of Transformations**

Think of transformations like medicine strength. You used the strongest medicine when you only needed a mild one.

| Strength | Transformation | Best For... |
| --- | --- | --- |
| **High** | **Log** (`np.log`) | Extreme exponential data (e.g., Salaries, House Prices). |
| **Medium** | **Square Root** (`np.sqrt`) | Count data or moderate skew (e.g., Call Duration, Ticket Counts). |
| **Optimal** | **Box-Cox** (`stats.boxcox`) | **The Gold Standard.** It calculates the exact strength needed automatically. |
